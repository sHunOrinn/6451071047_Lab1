```python
# ===== Câu 1 (a): Thêm tiêu đề các cột =====
print("Nguyễn Bình Minh - 6451071047")

import numpy as np
import pandas as pd
%matplotlib inline
from matplotlib import pyplot as plt

# File gốc không có dòng tiêu đề (nếu có thì đổi header=None thành header=0)

cols = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
df = pd.read_excel('tb_tracking.xlsx', header=None, names=cols)

display(df.head(10))
df.info()
```

```python
# ===== Câu 2 (b): Loại bỏ dữ liệu trước ngày 03/01/2018 =====
print("Nguyễn Bình Minh - 6451071047")

df['time'] = pd.to_datetime(df['time'], dayfirst=True)
print('Trước khi lọc:', len(df), 'dòng, sớm nhất', df['time'].min())
df = df[df['time'] >= pd.Timestamp(2018, 1, 3)].reset_index(drop=True)
print('Sau khi lọc :', len(df), 'dòng, sớm nhất', df['time'].min())
display(df.head())
```

```python
# ===== Câu 3 (c): Tách thời điểm đo trong ngày theo phút (minutes) =====
print("Nguyễn Bình Minh - 6451071047")

df['minutes'] = df['time'].dt.hour * 60 + df['time'].dt.minute

display(df[['time', 'minutes']].head(10))
```

```python
# ===== Câu 4 (d): Scatter plot co_level, humidity, temperature theo minutes =====
print("Nguyễn Bình Minh - 6451071047")

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(12, 12), sharex=True)
for ax, col, color in zip(axes, ['co_level', 'humidity', 'temperature'], ['tab:red', 'tab:blue', 'tab:orange']):
    ax.scatter(df['minutes'], df[col], s=4, alpha=0.5, color=color)
    ax.set_ylabel(col)
    ax.set_title(f'{col} theo minutes')
axes[-1].set_xlabel('minutes (phút trong ngày)')
plt.tight_layout()
plt.show()
```
